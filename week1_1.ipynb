```python
from scipy import stats
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('ggplot')
%matplotlib inline  
#% - магическая команда юпитера
```

```python
import numpy as np
```

```python

norm_rv = stats.norm(loc=0, scale=1)
```

```python
sample = norm_rv.rvs(1000)
sample[:10]
"""
rvs - выборка заданного размера
pdf - плотность нормального распределения d - Density
ppf - квантиль

cdf - кумулятивная функция распределения(x<X)
"""
```

```python
norm_rv.pdf(1)
norm_rv.cdf(1)
```

```python
x = np.linspace(-3,3,100)
x[:10]
```

```python
xc = np.linspace(-3,1,100)
xc[:10]
```

```python
pdf = norm_rv.pdf(x)
plt.plot(x, pdf)
plt.ylabel('$f(x)$')
plt.xlabel('$x$')
# $ - включает мат режим формул

plt.scatter([0, 1], [norm_rv.pdf(0),norm_rv.pdf(1)], color = 'green')
plt.text (0, norm_rv.pdf(0), f"P = {z_calc:.2f}" )
```

```python
pdf = norm_rv.pdf(x)
plt.plot(x, pdf)
plt.ylabel('$f(x)$')
plt.xlabel('$x$')
# $ - включает мат режим формул

plt.scatter([0, 1], [norm_rv.pdf(0),norm_rv.pdf(1)], color = 'blue')


xq = np.linspace(-3,1,100)
yq = norm_rv.pdf(xq)
plt.fill_between(xq, 0,yq, alpha = 0.3, color = 'blue')
plt.ylim(bottom=0)
plt.axvline(1, color = 'blue', linestyle = ':', lw = '1')
```

```python
cdf = norm_rv.cdf(x)
plt.plot(x, cdf)
plt.ylabel('$f(x)$')
plt.xlabel('$x$')
# $ - включает мат режим формул

plt.scatter([0, 1], [norm_rv.cdf(0),norm_rv.cdf(1)], color = 'blue')


xc = np.linspace(-3,1,100)
yc = norm_rv.cdf(xc)
plt.fill_between(xc, 0,yc, alpha = 0.5, color = 'blue')
plt.ylim(bottom=0)
plt.axvline(1, color = 'blue', linestyle = ':', lw = '1')
```

```python

pdf = norm_rv.pdf(x)
plt.plot(x, pdf)
plt.ylabel('$f(x)$')
plt.xlabel('$x$')
# $ - включает мат режим формул


z1 = 1
z2 = 2

plt.scatter([z1, z2], [norm_rv.pdf(z1),norm_rv.pdf(z2)], color = 'blue')




xq = np.linspace(z1,z2,100)
yq = norm_rv.pdf(xq)
plt.fill_between(xq, 0,yq, alpha = 0.2, color = 'blue')
plt.ylim(bottom=0)
plt.axvline(z1, color = 'blue', linestyle = ':', lw = '1')
plt.axvline(z2, color = 'blue', linestyle = ':', lw = '1')


z_calc = norm_rv.cdf(z2) - norm_rv.cdf(z1)
# Add text annotation
plt.text( (z1 + z2)/2, 0.1, f"P = {z_calc:.4f}" )
plt.title("Normal Distribution with Area Highlighted")
plt.grid(True)



qw = norm_rv.ppf(0.05)
xqw = np.linspace(-3, qw,100)
yqw = norm_rv.pdf(xqw)
plt.fill_between(xqw, 0,yqw, alpha = 0.2, color = 'green')
plt.text( (-3 + qw)/2, 0.1, f"P = {qw:.4f}" )
plt.ylim(bottom=0)

plt.show()
```

```python
x = np.linspace(0,100,1000)
norm_age = stats.norm(loc=32, scale=10)
pdf_age = norm_age.pdf(x)

plt.plot(x, pdf_age)
plt.ylabel('$f(x)$')
plt.xlabel('$x$')


for i in range(len(df)):
    x = df.iloc[i]["age"]
    y = norm_age.pdf(x)
    name = df.iloc[i]["name"]
    plt.scatter(x, y, color='blue')
    plt.text(x * 1.01, y + 0.001, name, fontsize=9)
    
mean_age = df["age"].mean()
median_age = df["age"].median()
std_age = df["age"].std()

stats_text = (
    f"Mean: {mean_age:.2f}\n"
    f"Median: {median_age:.2f}\n"
    f"Std Dev: {std_age:.2f}"
)

plt.text(
    x=plt.xlim()[1] * 0.7,      # 70% from the left (right side)
    y=plt.ylim()[1] * 0.7,     # 95% from the bottom (top)
    s=stats_text,
    fontsize=10,
    bbox=dict(facecolor='white', edgecolor='black')
)
```
